## **Conditional Statements**---> used when we want to execute certain blocks of code if they fullfil specified conditions. There are mainly three conditional statements in python. Mainly:
#
## 1. **if**--> simplest conditional statement, executes the code if given condition is satisfied.

## 2. **elif**--> stands for else-if and is generally used when checking for multiple statements/conditions in a block of code.

## 3. **else**--> used to print the block of code which corresponds to when none of the specified conditions are met







```python
#1(a). To check whether a given number is positive or not.
x = int(input("Enter a Number: "))
if x>0:
  print("The number entered is positive.")
else:
  print("The number entered is negative.")


#1(b). To check whether a given number is positive, negative or zero.
x = int(input("Enter a Number: "))
if x>0:
  print("The number entered is positive.")
elif x<0:
  print("The number entered is negative.")
else:
  print("The number entered is zero.")
```

```python
#2. To check whether a given number is even or odd.
a = int(input("Enter a Number: "))
if a%2==0:
  print("The entered number is even.")
else:
  print("The entered number is odd.")
```

```python
#3. To find the largest of two numbers
x = int(input("Enter the First no.: "))
y = int(input("Enter the Second no.: "))

if x>y:
  print(x,"is greater than",y)
elif x<y:
  print(y,"is greater than",x)
else:
  print("Both numbers are equal.")
```

```python
#4. To find the largest of three numbers
x = int(input("Enter the First no.: "))
y = int(input("Enter the Second no.: "))
z = int(input("Enter the Third no.: "))

if x>y and x>z:
  print(x,"is the largest number.")
elif y>x and y>z:
  print(y,"is the largest number.")
else:
  print(z,"is the largest number.")
```

```python
#5. To input marks of 5 subjects of a student, find the average of the marks and print the grade based on it.
p = int(input("Enter the marks of the First subject: "))
q = int(input("Enter the marks of the Second subject: "))
r = int(input("Enter the marks of the Third subject: "))
s = int(input("Enter the marks of the Fourth subject: "))
t = int(input("Enter the marks of the Fifth subject: "))

avg = (p+q+r+s+t)/5

if avg >= 90:
  print("Achieved Grade: A")
elif avg >= 80 and avg <90:
  print("Achieved Grade: B")
elif avg >= 70 and avg <80:
  print("Achieved Grade: C")
elif avg >= 60 and avg <70:
  print("Achieved Grade: D")
else:
  print("Achieved Grade: F")
```

```python
#6. Check whether a year is a leap year
y = int(input("Enter a year: "))
if(y%400==0) or (y%4==0 and y%100!=0):
  print("Given year is a leap year.")
else:
  print("Given year is not a leap year.")
```
